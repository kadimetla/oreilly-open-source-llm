# Running `vllm` as a server

In the previous notebooks, you have seen how `vllm` can work as a
(powerful) library. Most of the time, however, you will run `vllm`
as a separate server process.

This has several advantages:
* `vllm` exposes an OpenAI compatible API
* You can easily migrate existing software to use this API (if OpenAI has been used before)
* Many different programs can access the API
* You could work with load balancing if it becomes necessary

We will use the OpenAI client:

```python
from openai import OpenAI
```

An API key can be added, but we don't need it here

```python
openai_api_key = "EMPTY"
openai_api_base = "http://localhost:8000/v1"
client = OpenAI(
    api_key=openai_api_key,
    base_url=openai_api_base,
)
```

Run: `vllm serve cyankiwi/Qwen3.5-9B-AWQ-4bit --max-model-len 16384`

The chat completion can be used in the usual way:

```python
model = "cyankiwi/Qwen3.5-9B-AWQ-4bit"
```

```python
%%time
completion = client.chat.completions.create(
    model=model, 
    messages=[{ "role": "user",
                "content": "Explain O'Reilly online learning!" } ]
)
```

```python
completion
```

```python
from IPython.display import display, Markdown
display(Markdown(completion.choices[0].message.reasoning))
```

It is also possible to work with the more modern `responses` API:

```python
response = client.responses.create(model=model, 
                                   input="Explain O'Reilly online learning!?")
```

```python
response
```

```python
display(Markdown(response.output_text))
```
